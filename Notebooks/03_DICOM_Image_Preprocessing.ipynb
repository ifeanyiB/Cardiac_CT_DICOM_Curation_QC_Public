# 03 - DICOM Image Preprocessing

## Project
DICOM Extension: Medical Image Visualization and Preprocessing

## Objective
Explore and implement common preprocessing techniques for medical images using a CT DICOM image.

## Tasks
- Load and inspect the CT image
- Apply intensity normalization
- Resize the image
- Apply noise reduction
- Explore contrast enhancement
- Compare original and preprocessed images
- Understand the effects of preprocessing on medical image data

## 1. DICOM Image Loading and HU Conversion

This section loads the CT image from the DICOM file, extracts its pixel data and converts the raw pixel values into Hounsfield Units (HU).

The resulting image serves as the input for the subsequent preprocessing operations.

```python
# IMPORT REQUIRED LIBRARIES AND LOAD THE CT DICOM IMAGE

# Import libraries
import numpy as np
import matplotlib.pyplot as plt
import pydicom
from pathlib import Path

# Define project and data paths
project_path = Path.cwd().parent
data_path = project_path / "Data"

# Load the DICOM file
dicom_file = data_path / "IM-0001-0001.dcm"
dcm = pydicom.dcmread(dicom_file)

# Extract pixel data
pixel_array = dcm.pixel_array

# Convert pixel values to Hounsfield Units
ct_image = pixel_array * float(dcm.RescaleSlope) + float(dcm.RescaleIntercept)

# Display basic image information
print("Image shape:", ct_image.shape)
print("Data type:", ct_image.dtype)
print("Minimum HU:", ct_image.min())
print("Maximum HU:", ct_image.max())
```

## 2. CT Image Intensity Normalization

This section applies min-max normalization to transform the CT image intensity values into a range between 0 and 1.

Normalization standardizes the numerical intensity range and prepares the image for subsequent preprocessing operations.

```python
# NORMALIZE CT IMAGE INTENSITY VALUES TO A RANGE BETWEEN 0 AND 1

# Find minimum and maximum intensity values
min_value = ct_image.min()
max_value = ct_image.max()

# Apply min-max normalization
normalized_image = (ct_image - min_value) / (max_value - min_value)

# Display normalization information
print("Original minimum HU:", min_value)
print("Original maximum HU:", max_value)

print("\nNormalized image:")
print("Minimum value:", normalized_image.min())
print("Maximum value:", normalized_image.max())
print("Data type:", normalized_image.dtype)
```

```python
# VISUALIZE THE ORIGINAL CT IMAGE AND NORMALIZED IMAGE

# Create a figure with two subplots
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

# Display the original CT image
axes[0].imshow(ct_image, cmap="gray")
axes[0].set_title("Original CT Image (HU)")
axes[0].axis("off")

# Display the normalized image
axes[1].imshow(normalized_image, cmap="gray")
axes[1].set_title("Normalized CT Image (0–1)")
axes[1].axis("off")

# Display both images
plt.tight_layout()
plt.show()
```

## 3. CT Image Resizing

This section demonstrates spatial resizing of the CT image from 512 × 512 pixels to 256 × 256 pixels.

Resizing reduces image dimensions and computational requirements while preserving the overall image structure.

```python
# RESIZE THE CT IMAGE TO A SMALLER DIMENSION

# Import the resizing function
from skimage.transform import resize

# Resize the CT image to 256 × 256 pixels
resized_image = resize(
    ct_image,
    (256, 256),
    preserve_range=True,
    anti_aliasing=True
)

# Display image dimensions
print("Original image shape:", ct_image.shape)
print("Resized image shape:", resized_image.shape)
```

## 4. Gaussian Filtering for Noise Reduction

This section applies Gaussian filtering to the original CT image to demonstrate image smoothing and noise reduction.

The original and filtered images are compared visually, and their standard deviations are examined to assess the effect of smoothing.

```python
# APPLY GAUSSIAN FILTERING TO REDUCE IMAGE NOISE

# Import Gaussian filtering function
from scipy.ndimage import gaussian_filter

# Apply Gaussian filter to the CT image
smoothed_image = gaussian_filter(ct_image, sigma=1)

# Display image information
print("Original image shape:", ct_image.shape)
print("Smoothed image shape:", smoothed_image.shape)
print("Original image standard deviation:", ct_image.std())
print("Smoothed image standard deviation:", smoothed_image.std())
```

```python
# COMPARE THE ORIGINAL CT IMAGE WITH THE GAUSSIAN-FILTERED IMAGE

# Create two subplots
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

# Display original CT image
axes[0].imshow(ct_image, cmap="gray")
axes[0].set_title("Original CT Image")
axes[0].axis("off")

# Display Gaussian-filtered image
axes[1].imshow(smoothed_image, cmap="gray")
axes[1].set_title("Gaussian Filtered Image (sigma=1)")
axes[1].axis("off")

# Display comparison
plt.tight_layout()
plt.show()
```

## 5. Contrast Enhancement Using CLAHE

This section applies Contrast Limited Adaptive Histogram Equalization (CLAHE) to enhance local image contrast.

The enhanced image is compared with the original CT image to demonstrate the effects of local contrast enhancement.

```python
# APPLY CLAHE TO ENHANCE LOCAL IMAGE CONTRAST

# Import CLAHE function
from skimage.exposure import equalize_adapthist

# Apply CLAHE to the CT image
clahe_image = equalize_adapthist(
    normalized_image,
    clip_limit=0.03
)

# Display image information
print("Original image shape:", ct_image.shape)
print("CLAHE image shape:", clahe_image.shape)
print("CLAHE minimum value:", clahe_image.min())
print("CLAHE maximum value:", clahe_image.max())
```

```python
# COMPARE THE ORIGINAL CT IMAGE WITH THE CLAHE-ENHANCED IMAGE

# Create two subplots
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

# Display original CT image
axes[0].imshow(ct_image, cmap="gray")
axes[0].set_title("Original CT Image")
axes[0].axis("off")

# Display CLAHE-enhanced image
axes[1].imshow(clahe_image, cmap="gray")
axes[1].set_title("CLAHE Enhanced Image")
axes[1].axis("off")

# Display comparison
plt.tight_layout()
plt.show()
```

## 6. Comparative Evaluation of Preprocessing Techniques

This section brings together the original CT image and the different preprocessing outputs for visual comparison.

The comparison includes:

- Original CT image.
- Normalized image.
- Resized image.
- Gaussian-filtered image.
- CLAHE-enhanced image.

This provides an overview of how each preprocessing operation affects the image.

```python
# COMPARE DIFFERENT PREPROCESSING RESULTS

# Create a figure to compare the original image and preprocessing outputs
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

# Display original CT image
axes[0, 0].imshow(ct_image, cmap="gray")
axes[0, 0].set_title("Original CT Image (HU)")
axes[0, 0].axis("off")

# Display normalized image
axes[0, 1].imshow(normalized_image, cmap="gray")
axes[0, 1].set_title("Normalized Image")
axes[0, 1].axis("off")

# Display resized image
axes[0, 2].imshow(resized_image, cmap="gray")
axes[0, 2].set_title("Resized Image (256 × 256)")
axes[0, 2].axis("off")

# Display Gaussian-filtered image
axes[1, 0].imshow(smoothed_image, cmap="gray")
axes[1, 0].set_title("Gaussian Filtered Image")
axes[1, 0].axis("off")

# Display CLAHE-enhanced image
axes[1, 1].imshow(clahe_image, cmap="gray")
axes[1, 1].set_title("CLAHE Enhanced Image")
axes[1, 1].axis("off")

# Leave the final subplot empty
axes[1, 2].axis("off")

# Adjust spacing between plots
plt.tight_layout()

# Display the comparison
plt.show()
```
